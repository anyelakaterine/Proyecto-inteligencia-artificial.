```
# ==========================================================
# NOMBRE DEL CÓDIGO:
# HEURÍSTICA DISTANCIA MANHATTAN
#
# DESCRIPCIÓN:
# Calcula la distancia entre dos puntos en una cuadrícula.
# ==========================================================

def heuristica_manhattan(x1,y1,x2,y2):

    distancia = abs(x1-x2) + abs(y1-y2)

    return distancia

print("Resultado heurística Manhattan:",heuristica_manhattan(1,2,4,6))

# ==========================================================
# NOMBRE DEL CÓDIGO:
# HEURÍSTICA DISTANCIA EUCLIDIANA
#
# DESCRIPCIÓN:
# Calcula la distancia directa entre dos puntos.
# ==========================================================

import math

def heuristica_euclidiana(x1,y1,x2,y2):

    distancia = math.sqrt((x1-x2)**2 + (y1-y2)**2)

    return distancia

print("Resultado heurística Euclidiana:",heuristica_euclidiana(1,2,4,6))
```
