# Proyecto de Inteligencia Artificial

## Algoritmo A*, Heurísticas y Teoría de Juegos

Estudiante: Anyela Katerine Renteria Cuama

Este proyecto presenta una investigación sobre el algoritmo A*, el uso de heurísticas
y la teoría de juegos, junto con implementaciones prácticas utilizando Python.

## 1. Introducción

La inteligencia artificial utiliza diferentes técnicas para resolver problemas
complejos. Entre ellas se encuentran los algoritmos de búsqueda, las heurísticas
y los modelos de decisión estratégica.

Uno de los algoritmos más utilizados para encontrar rutas óptimas es el algoritmo
A*, el cual combina el costo real del camino recorrido con una estimación del
costo restante hacia el objetivo.

Además, las heurísticas permiten mejorar la eficiencia de los algoritmos de
búsqueda, mientras que la teoría de juegos permite analizar la toma de decisiones
entre diferentes agentes o jugadores.

## 2. Objetivos

### Objetivo general

Comprender el funcionamiento del algoritmo A*, las heurísticas y la teoría de
juegos mediante la implementación de ejemplos en Python.

### Objetivos específicos

- Investigar el funcionamiento del algoritmo A*
- Implementar el algoritmo en Python
- Visualizar el recorrido del algoritmo
- Analizar el uso de heurísticas
- Implementar ejemplos de teoría de juegos
